```python
import json
import os
from typing import Dict, Any, List

# Load the tournament results from the JSON file
def load_tournament_results(filename: str) -> Dict[str, Any]:
    """
    Load tournament results from a JSON file.
    
    Args:
        filename (str): Path to the JSON file containing tournament results
        
    Returns:
        Dict[str, Any]: Dictionary containing the tournament results data
    """
    with open(filename, 'r') as f:
        data = json.load(f)
    return data

# Load the tournament results
tournament_results = load_tournament_results("tournament_results_20250401_125033.json")

# Extract the list of model names from the tournament results
model_names: List[str] = tournament_results["tournament_metadata"]["model_names"]

# Extract the final ELO ratings from the tournament results
final_elo_ratings: Dict[str, float] = tournament_results["final_elo_ratings"]

# Create a DataFrame to display the final ELO ratings
import pandas as pd

# Create a DataFrame with model names and their final ELO ratings
elo_df = pd.DataFrame({
    "Model": list(final_elo_ratings.keys()),
    "Final ELO Rating": list(final_elo_ratings.values())
})

# Sort the DataFrame by ELO rating in descending order
elo_df = elo_df.sort_values(by="Final ELO Rating", ascending=False)

# Reset the index for cleaner display
elo_df = elo_df.reset_index(drop=True).round(0)
# Initialize a dictionary to track statistics for each model
model_stats = {model: {"time": 0, "messages": 0, "invalid_votes": 0} for model in model_names}

# Loop through each contest in the tournament results
for contest in tournament_results.get("contests", []):
    model_a = contest["model_a"]
    model_b = contest["model_b"]
    
    # Loop through each game in the contest
    for game in contest.get("games", []):
        # Extract the models playing as mafia and townsperson
        model_mafia = game["model_mafia"]
        model_townsperson = game["model_townsperson"]
        
        # Update statistics for the mafia model
        model_stats[model_mafia]["time"] += game.get("mafia_total_time", 0)
        model_stats[model_mafia]["messages"] += game.get("mafia_total_messages", 0)
        model_stats[model_mafia]["invalid_votes"] += game.get("mafia_invalid_votes", 0)
        
        # Update statistics for the townsperson model
        model_stats[model_townsperson]["time"] += game.get("townsperson_total_time", 0)
        model_stats[model_townsperson]["messages"] += game.get("townsperson_total_messages", 0)
        model_stats[model_townsperson]["invalid_votes"] += game.get("townsperson_invalid_votes", 0)

stats = pd.DataFrame(model_stats).transpose()
stats["latency"] = stats["time"] / stats["messages"]
stats = stats.round(1)
stats, elo_df



```
